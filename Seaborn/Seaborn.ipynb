```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
print(sns.get_dataset_names())
a=sns.load_dataset('tips')
a.columns # dataset ke column dekhne ke liye 
# pd.unique(a['time'])
# or
# print(a['time'].unique())

sns.scatterplot(a,x='total_bill',y='tip',hue='sex',size='size',style='smoker')
```

```python
sns.lineplot(a,x='total_bill',y='tip')
```

```python
dt=sns.load_dataset('tips')
plt.figure(figsize=(5,5)) # fro size and it work with in all plot
sns.relplot(data=dt,x='total_bill',y='tip',hue='day',size='size',kind='line',height=5)
sns.relplot(data=dt,x='total_bill',y='tip',hue='day',size='size',kind='scatter',height=7)
plt.show()
```

```python
plt.figure(figsize=(4,4))

plt.subplot(1,3,1)
pt1=np.array([23,44,22])
plt.pie(pt1)


plt.subplot(1,3,2)
pt2=np.array([67,76,88])
plt.pie(pt2)


plt.subplot(1,3,3)
pt3=np.array([23,84,92])
plt.pie(pt3)

plt.show()

plt.bar(dt['total_bill'], dt['tip'])
```

```python
sns.kdeplot(data=dt,x='tip',hue='sex',fill=True)
```

```python
hotel=sns.load_dataset('tips')
sns.boxplot(data=hotel,x='day',y='total_bill',hue='sex')

```

```python
sns.catplot(data=hotel,x='day',y='total_bill',hue='sex',kind='box')
```

```python
# combination of violinplot and catplot
sns.violinplot(data=hotel,x='day',y='total_bill',hue='sex')
sns.catplot(data=hotel,x='day',y='total_bill',hue='sex',kind='violin')
```

```python
# Barplot : Axis Level
sns.barplot(data=hotel,x='day',y='total_bill',estimator='min')
```

```python
sns.barplot(data=hotel,x='day',y='total_bill',estimator='max')
```

```python
sns.catplot(data=hotel,y='total_bill',kind='bar')
```

```python
a=sns.countplot(data=hotel,x='day')
a.bar_label(a.containers[0])
```
